### Q13. Write a Python program to count the number of vowels, consonants, digits and special characters in a string

```python
sentence = "Hello World"
count_vowels = 0
count_consonants = 0
count_digits = 0
count_special_characters = 0
for char in sentence:
    if char.lower() in "aeiou":
        count_vowels += 1
    elif char.isalpha():
        count_consonants += 1;
    elif char.isdigit():
        count_digits += 1;
    else:
        count_special_characters += 1;
print(sentence)
print(f"Number of vowels {count_vowels}");
print(f"Number of consonants {count_consonants}");
print(f"Number of digits {count_digits}");
print(f"Number of special characters {count_special_characters}");
```

### Q14. Write a Python program to count Uppercase, Lowercase, special characters, and numeric values in a given string.

```python
sentence = "Hello World! 123"
count_upper = 0
count_lower = 0
count_special_characters=0
count_numeric = 0

for character in sentence:
    if character.isupper():
        count_upper += 1
    if character.islower():
        count_lower += 1
    if not character.isalnum():
        count_special_characters += 1
    if character.isnumeric():
        count_numeric += 1

print(sentence)
print(f"Number of Uppercase values {count_upper}");
print(f"Number of Lowercase values {count_lower}");
print(f"Number of special characters values {count_special_characters}");
print(f"Number of numeric values {count_numeric}"); 
```

### Q15. Write a program to make a new string with all the consonants deleted from the string "Hello, have a good day".

### Q16. Write a Python program to remove the nth index character from a non-empty string.

```python
sentence = input("Enter a non-empty string: ")
index = int(input("Index for character to be removed: "))
sentence = ''.join(ch for i, ch in enumerate(sentence) if i != index)
print(sentence)
```

### Q39. Write a Python Program to find simple interest and compound interest

```python
'''
P - Principal - Original amount
R - Rate of simple interest (percentage per annum)
T/t - Time in Years
n = number of times interest is compounded per year
SI = P * R * T / 100
CI = P[(1 + r/n)^nt - 1]
'''
p = 5000
r = 8
t = 3
si = p*r*t/100
print(f"Simple Interest {si}")
n=1
#Compound interest compounded annually
ci = p*((1 + r/(n*100))**(n*t) - 1)
print(f"Compund Interest {ci:.2f}")
```

### Q40. Write a Python program to check whether a number is Prime or not

```python
import math
number = 17
isPrime = True
if(number < 2):
    isPrime = False
else:
    for i in range(2, math.isqrt(number) +1):
        if number%i == 0:
            isPrime = False
            break
    if isPrime :
        print(f"{number} is prime")
    else :
        print(f"{number} is not Prime")
```
