### Q27. How to sort a list in ascending and descending order without using the sort function?

```python
def sort(seq,asc=True):
    for i in range(len(seq)):
        for j in range (len(seq)-i-1):
            if (asc and seq[j] < seq[j+1]) or ((not asc) and seq[j] > seq[j+1]):
                seq[j] , seq[j+1] = seq[j+1], seq[j]
                
seq = [4,2,1,10,6,8,3]
print(seq)
sort(seq)
print(seq)
sort(seq,False)
print(seq)
```

### Q35. What is the difference between copy and deep copy?

change in nested element in a copy will update the original also however this will not happen in deep copy

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

# Modifying a nested element affects both
shallow.append(100)
shallow[0].append(99)
print(original)  # Output: [[1, 2, 99], [3, 4]]
print(shallow)   # Output: [[1, 2, 99], [3, 4]]
```

```python
import copy

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

# Modifying a nested element affects only the deep copy
deep.append(100)
deep[0].append(99)

print(original)  # Output: [[1, 2], [3, 4]]
print(deep)      # Output: [[1, 2, 99], [3, 4]]
```

### Q39. Write a function to reverse the elements within each tuple in a list of tuples.

tuples = [(1, 2), (3, 4), (5, 6)]

Output: [(2, 1), (4, 3), (6, 5)]

```python
tuples = [(1, 2), (3, 4), (5, 6)]
for i,sub_tuple in enumerate(tuples):
    tuples[i] = tuple(sorted(sub_tuple, reverse=True))
tuples
```

### Q40. Write a function to add corresponding elements of tuples in two lists of tuples

list1 = [(1, 2), (3, 4)]

list2 = [(5, 6), (7, 8)]

Output: [(6, 8), (10, 12)]

```python
list1 = [(1, 2), (3, 4)]
list2 = [(5, 6), (7, 8)]
list3 = []
for i in range(len(list1)):
    tmpLst = []
    for j in range(len(list1[i])):
        tmpLst.append((list1[i][j] + list2[i][j]))
    list3.append(tuple(tmpLst))
print(list3)
```
